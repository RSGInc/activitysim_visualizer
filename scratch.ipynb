```python
import glob
import math
import openmatrix as omx
import numpy as np

for path in glob.glob(r"C:\Users\wesley.darling\project_data\odot_skims\*.omx"):
    with omx.open_file(path, "r") as f:
        for name in sorted(f.list_matrices()):
            if name.startswith("KTW_CRT__") or name.startswith("WTK_CRT__"):
                m = np.asarray(f[name][:])
                finite = m[np.isfinite(m)]
                if finite.size == 0:
                    print(path, name, "no finite values")
                    continue
                nonzero = finite[finite != 0]
                print(
                    path,
                    name,
                    "shape=", m.shape,
                    "min=", float(finite.min()),
                    "max=", float(finite.max()),
                    "nonzero_count=", int(nonzero.size),
                    "zero_share=", float((finite == 0).sum()) / float(finite.size),
                )
```
